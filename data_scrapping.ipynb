```python

import pandas as pd
```

```python
df_2014_2015 = pd.read_html('https://fbref.com/en/comps/9/2014-2015/2014-2015-Premier-League-Stats', 
                            attrs = {"id" : "results2014-201591_overall" })[0]

df_2015_2016 = pd.read_html('https://fbref.com/en/comps/9/2015-2016/2015-2016-Premier-League-Stats', 
                            attrs = {"id" : "results2015-201691_overall" })[0]

df_2016_2017 = pd.read_html('https://fbref.com/en/comps/9/2016-2017/2016-2017-Premier-League-Stats', 
                            attrs = {"id" : "results2016-201791_overall" })[0]
                            
df_2017_2018 = pd.read_html('https://fbref.com/en/comps/9/2017-2018/2017-2018-Premier-League-Stats', 
                            attrs = {"id" : "results2017-201891_overall" })[0]

df_2018_2019 = pd.read_html('https://fbref.com/en/comps/9/2018-2019/2018-2019-Premier-League-Stats', 
                            attrs = {"id" : "results2018-201991_overall" })[0]

df_2019_2020 = pd.read_html('https://fbref.com/en/comps/9/2019-2020/2019-2020-Premier-League-Stats', 
                            attrs = {"id" : "results2019-202091_overall" })[0] 
                            
df_2020_2021 = pd.read_html('https://fbref.com/en/comps/9/2020-2021/2020-2021-Premier-League-Stats', 
                            attrs = {"id" : "results2020-202191_overall" })[0]   
                            
df_2021_2022 = pd.read_html('https://fbref.com/en/comps/9/2021-2022/2021-2022-Premier-League-Stats', 
                            attrs = {"id" : "results2021-202291_overall" })[0]
                            
df_2022_2023 = pd.read_html('https://fbref.com/en/comps/9/2022-2023/2022-2023-Premier-League-Stats', 
                            attrs = {"id" : "results2022-202391_overall" })[0]
                            
df_2023_2024 = pd.read_html('https://fbref.com/en/comps/9/2023-2024/2023-2024-Premier-League-Stats', 
                            attrs = {"id" : "results2023-202491_overall" })[0]
df_2024_2025 = pd.read_html('https://fbref.com/en/comps/9/2024-2025/2024-2025-Premier-League-Stats', 
                            attrs = {"id" : "results2024-202591_overall" })[0]
```

```python

# List of DataFrames 
dfs = [
    df_2014_2015, df_2015_2016, df_2016_2017, df_2017_2018, 
    df_2018_2019, df_2019_2020, df_2020_2021, df_2021_2022, 
    df_2022_2023, df_2023_2024, df_2024_2025
]

# Standardize column names
for df in dfs:
    df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

# Define expected columns (ensuring missing xG-related ones are included)
expected_columns = [
    "rk", "squad", "mp", "w", "d", "l", "gf", "ga", "gd", "pts", 
    "pts/mp", "attendance", "top_team_scorer", "goalkeeper", "notes",
    "xg", "xga", "xgd", "xgd/90"  # Expected goals columns missing in older data
]

# Add a season column to track the year
years = list(range(2014, 2025))
for df, year in zip(dfs, years):
    df["season"] = year

# Concatenate all seasons into one Pandas DataFrame
full_data = pd.concat(dfs, ignore_index=True)
```
